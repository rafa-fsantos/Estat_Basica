# Capítulo 1 - Preliminares

## [1.1] Introdução

A **Inferência estatística** é uma das partes da **Estatística**. A **estatística**, por sua vez faz parte da metodologia da **Ciência** que objetiva **coletar**, **reduzir**, **analisar** e **modelar** os dados, a partir da qual podemos fazer 
**inferências** sobre a população das qual os dados (amostras) forma obtidos.

O **método científico** para testar teorias ou hipóteses é resumido em quatro passos:

1. **Formulação** do problema ou teoria, ou teste de uma hipótese;
2. **Coleta** de informação relevante, que pode ser feito por meio de um experimento ou observação de fenomênos ou variáveis de interesse;
3. O **resultados** da coleta são utilizados para obter conclusões, mesmo que não definitivas;
4. Se for necessário, repete-se a **coleta** e a obtenção de **resultados**, ou reformular as **hipóteses**.

Na **Análise Exploratória de Dados (AED)** o foco é na redução, análise e interpretação dos dados. Com essa abordagem buscamos obter informações para a construção de **modelos** que serão utilizados na fase posterior, a **Análise Confirmatória de 
Dados** ou **Inferência Estatística**.

A **Análise Exploratória de Dados** é uma corrente moderna liderada por **Tukey (1977)** que utiliza técnicas gráficas em conjunto com o cálculo das medidas de posição e variabilidade, como por exemplo, média e variância.

## [1.2] Modelos

Em uma **análise de dados** buscamos uma **regularidade** ou **padrão**, ou seja, um **modelo** que seja adequado ou não, para a representação dos dados.

De modo esquemático:
$$DADOS = MODELO + RESÍDUOS$$ **ou**, $$D = M + R$$

O **Modelo (M)** é a **parte suave** (regular ou prevísivel), enquanto que a parte de **Resíduos (R)** é a **parte aleátoria**, a qual é uma parte fundamental do trabalho estatístico.

Basicamente são os **resíduos** que nos dizem se o **modelo** é adequado ou não e esperamos que os **resíduos** não contenham **suavidade**.

## [1.3] Estatística e ciência de dados

**Ciência de Dados** ou **Data Science** é uma mescla entre **Estatística** e **Computação**. É um novo paradigma que une formação em **estatística** e técnicas computacionais em linguagens de programação como **R** e **Python**, além de **redes 
neurais**.

Nesta nova área, termos como **Aprendizado com Estatística (*Statistical Learning*)** e **Aprendizado de Máquina (*Machine Learning, ML*)** passaram a ser utilizados e incluem por sua vez, **Aprendizado Supervisionado** e **Aprendizado Não 
Supervisionado**.

### [1.3.1] Aprendizado supervisionado

Inclui técnicas de **Regressão** e **Classificação** que fazem parte de disciplinas como **Análise de Regressão**, **Modelos Lineares**, **Modelos Lineares Gereralizados** e **Análise Multivariada**.

### [1.3.2] Aprendizado não supervisionado

Compreende técnicas como **Análise de Agrupamentos**, **Análise de Componentes Principais**, **Análise Fatorial** e **Análise de Componentes Independentes**. São técnicas da disciplina de **Análise Multivariada** e as três últimas são técnicas de redução de dimensionalidade dos dados, enquanto que a **Análise de Agrupamentos** é uma técnica de **clusterização**, ou agrupamento de dados, com base em características similares.

### [1.3.3] Dados e megadados

Dados normalmente são dispostos na forma de uma tabela $n \times p$ com $n$ **amostras** e $p$ **variáveis**.

Podem ser estruturados:

    a. grande número de amostras (n) e pequeno número de variáveis (p) - small data;
    b. pequeno número de amostras (n) e grande número de variáveis (p) - high dimension;
    c. grande número de unidades amostrais e grande número de variáveis (p) - big data.
