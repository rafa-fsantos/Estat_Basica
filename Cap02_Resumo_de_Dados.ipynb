```python
# Importar pacotes/bibliotecas
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
# Importar dados
tab21 = pd.read_csv('raw/tabela2_1.csv', sep=';')

# Equivalente em R
# tab21 <- read.table('tabela2_1.csv', dec=',', sep=';', h=T)
```

```python
# Visualizar uma amotra do dataset (topo)
tab21.head(10)
```

```python
# Criar a tabela de frequências
# Define a ordem desejada para a variável qualitativa ordinal
ordem = ['ensino fundamental',
         'ensino médio',
         'superior']

# Calcula a frequência absoluta (ni) de cada categoria
# e reorganiza o resultado conforme a ordem definida
frequencia = tab21['grau_instrucao'].value_counts().reindex(ordem)

# Calcula a frequência relativa (fi), em proporção,
# e reorganiza conforme a ordem definida
proporcao = tab21['grau_instrucao'].value_counts(normalize=True).reindex(ordem)

# Converte a proporção para porcentagem
porcentagem = proporcao * 100

# Cria a tabela de distribuição de frequências
dist_frequencias = pd.DataFrame({'Frequência (ni)': frequencia,
                                 'Proporção (fi)': round(proporcao, 2),
                                 'Porcentagem (100 fi)': round(porcentagem, 2)})

# Adiciona uma linha com os totais das colunas
dist_frequencias.loc['Total'] = [dist_frequencias['Frequência (ni)'].sum(),
                                 dist_frequencias['Proporção (fi)'].sum(),
                                 dist_frequencias['Porcentagem (100 fi)'].sum()]

# Exibe a tabela final
dist_frequencias
```
