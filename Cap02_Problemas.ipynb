### [1] Escalas de medidas.

A seguir descrevemos outros possíveis critérios para classificar variáveis, em função da escala adotada. Observe a similaridade com a classificação apresentada anteriormente. Nossas observações são resultados
de medidas feitas sobre os mesmos elementos de uma populção. Existem quatro escalas de medidas que podem ser consideradas:

**Escala nominal.**
Nesta escala, somente podemos afirmar que uma medida, é diferente ou não de outra, e ela é usada para categorizar indivíduos de uma população. Um exemplo é o sexo de um indíviduo. Para caa categoria associamos
um numeral diferente (letra ou número). Por exemplo, no caso do sexo, podemos associar as letras M (masculino) e F (feminino) ou 1 (masculino) e 2 (feminino). Não podemos realizar operações aritméticas aqui e 
uma medida de posição apropriada é a moda. (As medidas citadas nesse problema, como a média, mediana e moda, são definidas no Capítulo 3.).

**Escala ordinal.**
Aqui, podemos dizer que uma medida é diferente e maior do que outra. Temos a situacdo anterior, mas as categorias são ordenadas, e a ordem dos numerais associados ordena as categorias. Por exemplo, a classe 
socioeconômica de um individuo pode ser baixa (1 ou X), média (2 ou Y) e alta (3 ou Z). Transformações que preservam a ordem não alteram a estrutura de uma escala ordinal. No exemplo acima, podemos 
representar as categorias por 1, 10 e 100 ou A, L e Z. Medidas de posição apropriadas são a mediana e a moda.

**Escala intervalar.**
Nesta escala, podemos afirmar que uma medida é igual ou diferente, maior e quanto maior do que outra. Podemos quantificar a diferenga entre as categorias da escala ordinal. Necessitamos de uma origem 
arbitrária e de uma unidade de medida. Por exemplo, considere a temperatura de um indivíduo, na escala Fahrenheit. A origem é 0° F e a unidade é 1° F. Transformações que preservam a estrutura dessa escala são 
do tipo $y = ax + b, a > 0$. Por exemplo, a transformação $y = 5/9 (x - 32)$ transforma graus Fahrenheit em centígrados. Para essa escala, podemos fazer operações aritméticas, assim a média, mediana e moda 
são medidas de posição apropriadas.

**Escala razão.**
Dadas duas medidas nessa escala, podemos dizer se são iguais, ou se uma é diferente, maior, quanto maior e quantas vezes a outra. A diferença com a escala intervalar é que agora existe um zero absoluto. A 
altura de um individuo é um exemplo de medida nessa escala. Se ela for medida em centímetros (cm), 0 cm é a origem e 1 cm é a unidade de medida. Um indíviduo com 190 cm é duas vezes mais alto do que um 
indíviduo com 95 cm, e esta relação continua a valer se usarmos 1 m como unidade. Ou seja, a estrutura da escala razão não é alterada por transformações da forma $y = cx, c > 0$. Por exemplo, $y = x/100$
transforma cm em m. As estatísticas apropriadas para a escala intervalar são também apropriadas para a escala razão.

### [2] Usando os dados da Tabela 2.1, construa a distribuição de frequências das variáveis:

(a) Estado civil.  
(b) Região de procedência.  
(c) Número de filhos dos empregados casados.  
(d) Idade.

```python
# Carrega o arquivo CSV 'tabela2_1.csv' localizado na pasta 'raw'.
# O separador utilizado no arquivo é ponto e vírgula (;).
tab21 = pd.read_csv('raw/tabela2_1.csv', sep=';')

# Exibe as primeiras 5 linhas do DataFrame para inspeção inicial.
tab21.head(10)
```

```python
# Proposta de solução para: (a) Estado civil
# Calcula a frequência absoluta de cada categoria da variável 'estado_civil'
freq_estcivil = tab21['estado_civil'].value_counts()

# Calcula a frequência relativa (proporção) de cada categoria
prop_estcivil = tab21['estado_civil'].value_counts(normalize=True)

# Converte as proporções em porcentagens
perc_estcivil = prop_estcivil * 100

# Cria uma tabela consolidando frequência absoluta,
# proporção e porcentagem de cada categoria
tab_freq_estcivil = pd.DataFrame({'Frequência (ni)': freq_estcivil,
                                  'Proporção (fi)': round(prop_estcivil, 2),
                                  'Porcentagem (100 * fi)': round(perc_estcivil, 2)})

# Adiciona uma linha com os totais das colunas
tab_freq_estcivil.loc['Total'] = [tab_freq_estcivil['Frequência (ni)'].sum(),
                                    tab_freq_estcivil['Proporção (fi)'].sum(),
                                    tab_freq_estcivil['Porcentagem (100 * fi)'].sum()]

# Exibe a tabela de distribuição de frequências
tab_freq_estcivil
```

```python
# Proposta de solução para: (b) Região de procedência

# Define a ordem desejada para apresentação das categorias
ordem = ['capital', 'interior', 'outra']

# Calcula a frequência absoluta de cada categoria
# e reorganiza os resultados conforme a ordem definida
freq_regproc = tab21['reg_procedencia'].value_counts().reindex(ordem)

# Calcula a frequência relativa (proporção) de cada categoria
# e reorganiza os resultados conforme a ordem definida
prop_regproc = tab21['reg_procedencia'].value_counts(normalize=True).reindex(ordem)

# Converte as proporções em porcentagens
perc_regproc = prop_regproc * 100

# Cria uma tabela contendo frequência absoluta,
# proporção e porcentagem de cada categoria
tab_freq_regproc = pd.DataFrame({'Frequência (ni)': freq_regproc,
                                 'Proporção (fi)': round(prop_regproc, 2),
                                 'Porcentagem (100 * fi)': round(perc_regproc, 2)})

# Adiciona uma linha de totais ao final da tabela
tab_freq_regproc.loc['Total'] = [tab_freq_regproc['Frequência (ni)'].sum(),
                                 tab_freq_regproc['Proporção (fi)'].sum(),
                                 tab_freq_regproc['Porcentagem (100 * fi)'].sum()]

# Exibe a tabela de distribuição de frequências
tab_freq_regproc
```

```python
# Proposta de solução para: (c) Número de filhos dos empregados casados.  

freq_empregcasad = tab21.groupby('estado_civil')['n_filhos'].value_counts(dropna=False).sort_index()
prop_empregcasad = tab21.groupby('estado_civil')['n_filhos'].value_counts(normalize=True, dropna=False).sort_index()
perc_empregcasad = prop_empregcasad * 100

tab_freq_empregcasad = pd.DataFrame({'Frequência (ni)': freq_empregcasad,
                                     'Proporção (fi)': round(prop_empregcasad, 2),
                                     'Porcentagem (100 * fi)': round(perc_empregcasad, 2)})

tab_freq_empregcasad.loc['Total'] = [tab_freq_empregcasad['Frequência (ni)'].sum(),
                                     tab_freq_empregcasad['Proporção (fi)'].sum(),
                                     tab_freq_empregcasad['Porcentagem (100 * fi)'].sum()]
tab_freq_empregcasad
```
